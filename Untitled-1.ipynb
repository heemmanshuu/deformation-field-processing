```python
import matplotlib.pyplot as plt
import numpy as np

# Grid sizes and times
times = np.array([6, 9, 20, 33, 60, 70, 158, 276, 488, 18471])
N = np.array([25, 49, 81, 121, 169, 225, 289, 361, 441, 5824])


# Plotting
plt.figure(figsize=(10, 6))
plt.plot(N, times, 'o-', label='Actual Time')
#plt.plot(N**2, times, 'r--', label='N² (scaled)')
#plt.plot(N22**3, times, 'g--', label='N³ (scaled)')

plt.xlabel('Grid Size (N × N)')
plt.ylabel('Time (seconds)')
plt.title('Time vs Grid Size (with N² and N³ comparison)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
